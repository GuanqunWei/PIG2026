## 相关知识

### 1. 中值滤波 Median Filter

中值滤波是一种非线性滤波方法。它用滤波窗口中像素灰度值的中位数替代中心像素值。由于椒盐噪声通常表现为极大值或极小值，中位数对这些离群值不敏感，因此中值滤波适合去除椒盐噪声。

补全时应思考：如何取出中心像素附近的邻域？如何根据核形状选择参与计算的像素？如何计算中位数？

### 2. 均值滤波 Mean Filter

均值滤波是一种线性平滑滤波方法。它用滤波窗口内像素的平均值替代中心像素值，可以抑制随机噪声，但也容易模糊图像边缘。

补全时应思考：均值滤波与中值滤波的流程很相似，不同点只在于最后的统计量。

### 3. 双边滤波 Bilateral Filter

双边滤波是一种边缘保持滤波方法。它同时考虑空间距离和像素值相似性：位置越近、灰度越接近的像素权重越大。

补全时应思考：空间权重如何由像素坐标距离得到？灰度权重如何由像素值差异得到？最终如何做加权平均？
