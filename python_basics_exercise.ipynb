## 2. Conda 是什么

- 不同项目可能依赖不同版本的第三方库。
- 独立环境可以减少包版本冲突。
- 便于在教学、实验和项目开发中保持环境清晰。

常见选择：

- Anaconda：包含较多常用科学计算工具，安装包较大。

常见用途：

## 8. 经典第三方库：OpenCV 与 cv2

OpenCV 是常用的计算机视觉和图像处理库，在 Python 中通常导入为 `cv2`。

常见用途：

注意：OpenCV 默认读取和处理图片时通常使用 BGR 颜色顺序，而 Matplotlib 显示图片时通常使用 RGB 颜色顺序，所以经常需要用 `cv2.cvtColor(image, cv2.COLOR_BGR2RGB)` 进行转换。

````python
# OpenCV / cv2 基础示例
try:
    import cv2
    import numpy as np
    import matplotlib.pyplot as plt
except ImportError as error:
    print("当前环境未安装 OpenCV、NumPy 或 Matplotlib。")
    print("可尝试执行：pip install opencv-python numpy matplotlib")
    print(error)
else:
    # 创建一张黑色图片。OpenCV 图片通常是 height x width x channels。
    image = np.zeros((200, 300, 3), dtype=np.uint8)

    # OpenCV 颜色顺序是 BGR，不是 RGB。
    cv2.rectangle(image, (30, 40), (140, 160), (255, 0, 0), thickness=3)   # 蓝色矩形
    cv2.circle(image, (220, 100), 45, (0, 255, 0), thickness=-1)           # 绿色实心圆
    cv2.putText(image, "cv2", (95, 190), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

    # 转成 RGB 后再用 plt 显示。
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(6, 4))
    plt.imshow(image_rgb)
    plt.title("OpenCV Drawing Example")
    plt.axis("off")
    plt.show()
````

## 13. 条件语句

条件语句用于根据不同情况执行不同代码。基本结构如下：

```python
if 条件1:
    条件1成立时执行
elif 条件2:
    条件2成立时执行
else:
```

Python 通过缩进表示代码块，因此缩进非常重要。

````python
score = 86

if score >= 90:
    level = "A"
elif score >= 80:
    level = "B"
elif score >= 60:
    level = "C"
else:
    level = "D"

print("level =", level)
````

## 15. 函数

基本结构：


## 16. Python 中的缩进与对齐

Python 使用缩进表示代码块。常见需要缩进的结构包括：

- `if`、`elif`、`else`
- `for`、`while`
- `def` 函数体
- `class` 类体

缩进规则：

1. 同一个代码块中的语句必须保持相同缩进。
2. 通常使用 4 个空格作为一级缩进。
3. 不建议混用 Tab 和空格。
4. 冒号 `:` 后面的下一行通常需要缩进。

缩进错误会导致程序无法运行，或者导致代码逻辑与预期不一致。

````python
# 缩进示例：if 语句中的代码块
score = 85

if score >= 60:
    print("及格")
    print("继续保持")
else:
    print("不及格")
    print("需要努力")

print("这行代码不属于 if 或 else，它总会执行。")
````

## 17. def 函数进阶

前面已经介绍过函数的基本写法。这里进一步说明函数中常见的几个概念：

- 参数：函数接收的输入。
- 返回值：函数计算后交回给调用者的结果。
- 默认参数：如果调用函数时没有传入某个参数，就使用默认值。
- 局部变量：在函数内部定义的变量，通常只在函数内部有效。

函数可以帮助我们把重复代码封装起来，让程序更清晰。

````python
def greet(name, message="你好"):
    """返回一条问候语。

    参数：
    - name: 被问候者姓名。
    - message: 问候内容，默认值为 "你好"。
    """
    return f"{message}，{name}！"


print(greet("小明"))
print(greet("Alice", "Hello"))
````

````python
def calculate_total(price, count, discount=1.0):
    """计算商品总价。

    参数：
    - price: 单价。
    - count: 数量。
    - discount: 折扣，默认 1.0 表示不打折。

    返回：
    - 商品总价。
    """
    total = price * count * discount
    return total


print(calculate_total(10, 3))
print(calculate_total(10, 3, 0.8))
````

## 18. class 类与对象

`class` 用于定义类。类可以理解为一种“模板”，对象是根据这个模板创建出来的具体实例。

例如，“学生”可以是一个类，每个具体学生就是一个对象。

类中常见概念：

- `class Student:`：定义一个学生类。
- `__init__`：初始化方法，创建对象时自动执行。
- `self`：表示当前对象本身。
- 属性：对象保存的数据，例如姓名、年龄、分数。
- 方法：定义在类中的函数，表示对象可以执行的行为。

类适合把相关数据和操作组织在一起。

````python
class Student:
    """学生类示例。"""

    def __init__(self, name, score):
        """初始化学生对象。"""
        self.name = name
        self.score = score

    def is_passed(self):
        """判断学生是否及格。"""
        return self.score >= 60

    def description(self):
        """返回学生信息描述。"""
        return f"{self.name} 的成绩是 {self.score}"


student = Student("Li Hua", 85)
print(student.name)
print(student.score)
print(student.is_passed())
print(student.description())
````
