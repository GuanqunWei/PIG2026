## 1. 相关背景知识

### 1.1 空间域与频率域

灰度图像可以看成一个二维函数：

$$
f(x,y)
$$

其中 `x` 和 `y` 表示像素位置，函数值表示灰度强度。

傅里叶变换可以把图像从 **空间域** 转换到 **频率域**。频率域中的每个复数系数表示一种二维正弦/余弦波成分：

1. 低频成分：表示图像中变化缓慢的部分，例如整体亮度和大块平滑区域。
2. 高频成分：表示图像中变化剧烈的部分，例如边缘、纹理和噪声。

### 1.2 二维离散傅里叶变换

对于大小为 `M × N` 的图像，二维 DFT 定义为：

$$
F(u,v)=\sum_{x=0}^{M-1}\sum_{y=0}^{N-1} f(x,y)
e^{-j2\pi(\frac{ux}{M}+\frac{vy}{N})}
$$

其中：

1. `f(x,y)` 是空间域图像。
2. `F(u,v)` 是频率域复数系数。
3. `u, v` 是频率索引。
4. `j` 是虚数单位。

### 1.3 二维离散傅里叶反变换

从频率域还原图像的 IDFT 为：

$$
f(x,y)=\frac{1}{MN}\sum_{u=0}^{M-1}\sum_{v=0}^{N-1} F(u,v)
e^{j2\pi(\frac{ux}{M}+\frac{vy}{N})}
$$

如果 DFT 和 IDFT 实现正确，反变换后的图像应该与原图非常接近。
